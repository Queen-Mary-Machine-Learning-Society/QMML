```python
dataset = pd.read_csv("../src/easy_dataset.csv")
```

```python
dataset.info()
```

```python
dataset.describe()
```

```python
dataset = np.asarray(dataset)
```

```python
dataset.shape
```

```python
inputs = dataset[:, :15]
targets = dataset[:, 15] + 100_000
```

```python
X = inputs
Y = targets
```

```python
B = np.linalg.inv(X.T @ X) @ X.T @ Y
```

```python
B
```

```python
Y_hat = X @ B
```

```python
mse = np.mean((Y_hat - Y)**2)
```

```python
print(f"The MSE is: {mse}")
```
