```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('src/easy_dataset.csv')
```

```python
df = df.to_numpy()
```

```python
df
```

```python
X = df[:, :15]
```

```python
column_ones = np.ones((X.shape[0], 1))
```

```python
X = np.concat([column_ones, X], axis=-1)
```

```python
y = df[:, 15]
```

```python
# Y = B @ X
# B_hat = (X.T @ X)^-1 @ X.T @ y


# y = B0 + B1x1 + B2x2 + B3x3 + ... + Bnxn
```

```python
b_hat = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
np.mean((y - X @ b_hat)**2)
```

```python
b_init = np.ones((16, 1))
```

```python
np.mean((y - X @ b_init)**2)
```
