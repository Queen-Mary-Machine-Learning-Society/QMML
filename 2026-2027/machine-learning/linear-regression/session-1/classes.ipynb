## Python Classes

A class in Python is a way to define a new kind of thing in your program. It describes what that thing knows (its data) and what it can do (its actions), all in one organized structure.

You can think of a class as a custom-made type you invent yourself, like creating a new word in the language of your program that represents a concept, complete with its own properties and behaviors.

A class allows you to write clean and reusable code

```python
# This is how a class stores data

class Dog:
    def __init__(self, age, breed):
        self.age = age
        self.breed = breed
```

```python
# This is how you create an instance of the class

dog = Dog(22, "Pug")
```

```python
dog.age, dog.breed
```

```python
# Classes can have methods (functions) that interact with the data

class Car:
    def __init__(self, brand, km, prev_owners):
        self.brand = brand
        self.km = km
        self.prev_owners = prev_owners
    
    def compute_km_per_owner(self):
        return self.km / self.prev_owners
    
    def get_car_description(self):
        description = f"Car: {self.brand} \n Km: {self.km} \n This car has {self.prev_owners} previous owner(s)."
        print(description)
        return description
```

```python
seat = Car("Seat", 200_000, 5)
```

```python
seat.compute_km_per_owner()
```
