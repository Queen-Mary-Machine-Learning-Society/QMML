```python
import torch
import torch.nn as nn
import numpy as np
```

```python
class NeuralNet(nn.Module):
    def __init__(self, in_dim: int, out_dm: int, hidden_dim: int):
        super().__init__()

        self.nn = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.ReLU(),

            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),

            nn.Linear(hidden_dim, out_dm)
        )

    def forward(self, x):
        return self.nn(x)
```

```python
# Let's create a synthetic dataset to do a forward pass

X = np.random.randn(1000, 4)
y = np.random.randn(1000, 1)
```

```python
X_t = torch.tensor(X, dtype=torch.float32)
y_t = torch.tensor(y, dtype=torch.float32)
```

```python
neural_net = NeuralNet(in_dim=X_t.shape[1], out_dm=y_t.shape[1], hidden_dim=8)
```

```python
predictions = neural_net.forward(X_t)
```

```python
mse = torch.mean((y_t-predictions)**2)
print(f"The mse of the untrained network is: {mse}")
```

```python
# we can observe the parameters of the model

for p in neural_net.parameters():
    print(p)
```
