# Essential Components of the TWS API
### API classes EClient and EWrapper
EClient is used for outgoing messages which are sent from the API program to TWS or the IB Gateway.

EWrapper is used to handle incoming messages from the Interactive Brokers server through TWS or IB Gateway."

```python
from ibapi.client import EClient
from ibapi.wrapper import EWrapper
from ibapi.contract import Contract
import threading
import time
import datetime
```

```python
# Importing EClient
from ibapi.client import *
# Importing EWrapper
from ibapi.wrapper import *
import time
import threading

# Create a class to combine the EClient and EWrapper modules
class TestApp(EClient, EWrapper):
  def __init__(self):
    EClient.__init__(self, self)
    self.orderId = 0

# Create the EWrapper object to receive order Ids, prevent duplicate values
  def nextValidId(self, orderId):
    self.orderId = orderId

  def nextId(self):
    self.orderId += 1
    return self.orderId

  def currentTime(self, time):
    print(time)

  def error(self, reqId, errorCode, errorString, advancedOrderReject=""):
    print(f"reqId: {reqId}, errorCode: {errorCode}, errorString: {errorString}, orderReject: {advancedOrderReject}")

app = TestApp()
app.connect("127.0.0.1", 7497, clientId=0) # 7497: Paper Trading, 7496: Live Trading
threading.Thread(target=app.run).start()
time.sleep(1)

# for i in range(0,5):
#   print(app.nextId())
app.reqCurrentTime()
```

```python

class TestApp(EClient, EWrapper):
    def __init__(self):
        EClient.__init__(self, self)
        self.data = []  # store retrieved bars here

    def nextValidId(self, orderId):
        self.orderId = orderId

    def error(self, reqId, errorCode, errorString, advancedOrderReject=""):
        if errorCode not in [2104, 2106, 2158]:
            print(f"reqId: {reqId}, errorCode: {errorCode}, errorString: {errorString}")

    # called for each bar of historical data
    def historicalData(self, reqId, bar):
        print(f"Time: {bar.date}, Open: {bar.open}, High: {bar.high}, Low: {bar.low}, Close: {bar.close}, Volume: {bar.volume}")
        self.data.append(bar.__dict__)

    # called when all bars have been received
    def historicalDataEnd(self, reqId, start, end):
        print(f"\n✅ Data received from {start} to {end}")
        self.disconnect()


# --- connect ---
app = TestApp()
app.connect("127.0.0.1", 7497, clientId=6)

thread = threading.Thread(target=app.run, daemon=True)
thread.start()

time.sleep(1)

# --- define contract ---
contract = Contract()
contract.symbol = "AAPL"
contract.secType = "STK"
contract.exchange = "NYSE" # Can change to "SMART", "NASDAQ", "LSE", etc.
contract.currency = "USD"

# --- request historical data ---
# Parameters:
#   reqId,
#   contract,
#   endDateTime ("" = now),
#   durationStr ("1 D", "1 W", "1 M", "1 Y", etc.)
#   barSizeSetting ("1 min", "5 mins", "1 hour", "1 day", etc.)
#   whatToShow ("TRADES", "MIDPOINT", "BID", "ASK"),
#   useRTH (1=regular trading hours only, 0=all hours),
#   formatDate (1=YYYYMMDD HH:MM:SS, 2=epoch)
#   keepUpToDate (False = one-time snapshot)
#   chartOptions ([])
app.reqHistoricalData(
    reqId=1,
    contract=contract,
    endDateTime="",
    durationStr="1 D",
    barSizeSetting="5 mins",
    whatToShow="TRADES",
    useRTH=1,
    formatDate=1,
    keepUpToDate=False,
    chartOptions=[]
)

# wait for data
time.sleep(10)

# convert to DataFrame (optional)
import pandas as pd
df = pd.DataFrame(app.data)
print(df.head())
```

### historical_market_data.py

```python
from ibapi.client import *
from ibapi.wrapper import *
import datetime
import time
import threading

port = 7497


class TestApp(EClient, EWrapper):
    def __init__(self):
        EClient.__init__(self, self)
        self.orderId = 0

    def nextValidId(self, orderId: OrderId):
        self.orderId = orderId

    def nextId(self):
        self.orderId += 1
        return self.orderId

    def error(self, reqId, errorCode, errorString, advancedOrderReject=""):
        print(f"reqId: {reqId}, errorCode: {errorCode}, errorString: {errorString}, orderReject: {advancedOrderReject}")

    def historicalData(self, reqId, bar):
        print(reqId, bar)

    def historicalDataEnd(self, reqId, start, end):
        print(f"Historical Data Ended for {reqId}. Started at {start}, ending at {end}")
        self.cancelHistoricalData(reqId)

app = TestApp()
app.connect("127.0.0.1", port, 3)
threading.Thread(target=app.run).start()
time.sleep(1)

mycontract = Contract()
mycontract.symbol = "AAPL"
mycontract.secType = "STK"
mycontract.exchange = "SMART"
mycontract.currency = "USD"

app.reqHistoricalData(app.nextId(), mycontract, "20240523 16:00:00 US/Eastern", "1 D", "1 hour", "TRADES", 1, 1, False, [])
```
